# 🧮 **Embeddings**, getting started

The representation of numerical data for computers is very simple, but how are more complex forms of data handled? **How do I represent words in a numerical form?** And more importantly, **how do I retain their meaning?**


An **embedding** is a numerical representation of a piece of information. Specifically, an embedding is a dense vector representation of data points in a lower-dimensional space, where similar data points are closer together, facilitating efficient computation and meaningful analysis.

## A dense vector? A ... what?

A dense vector is simply a fancy way of saying that **we want to place our information on a Cartesian plane**. Why? Because this is the simplest solution that allows us humans **to understand the "meaning"** of a specific embedding and at the same time allows a computer to have a **representation in numerical form** that it understands!

Visually these embeddings should look something like this:

![Embeddings](https://miro.medium.com/v2/resize:fit:2000/1*SYiW1MUZul1NvL1kc1RxwQ.png)


As evident **similar concepts are close to each other** in the Cartesian plane: we humans understand this form and likewise computers have their own numerical representation in the form of coordinates!

> Actually this representation is not new at all, we use it with text but it can be used for any complex data such as documents, images, graphs, and even as pieces of knowledge.

### Okay so where do we start?

```python
from gensim.models import Word2Vec
import time
import numpy as np

# Example sentences for training the Word2Vec model
sentences = [["this", "is", "a", "sentence"], ["this", "is", "another", "sentence"]]

# Train Word2Vec model
default_model = Word2Vec(sentences, min_count=1)
```

Extract the words and thier corresponding vector

```python
# Get all word vectors and corresponding words
word_vectors = default_model.wv
word_vecs = [word_vectors[word] for word in word_vectors.index_to_key]
```

```python
# This is the embedding of the first word!
word_vecs[0]
```

### Uhm, many numbers?

In the figure above it is evident how there are 3 coordinates (`[x, y, z]`). In reality, however, embeddings have many more dimensions because obviously 3 is too few to represent the true meaning of something.

To give an idea, the simplest representations exceed 50 dimensions, state of the art models (even GPT 4 chat for example) even use more than 2000 dimensions! 🤯

```python
print(f'The dimension of this embedding is {len(word_vecs[0])}')
```

#### Yes, but with all these dimensions, how do I represent them on a Cartesian plane to observe them?

That is why there are [dimensionality reduction techniques](https://en.wikipedia.org/wiki/Dimensionality_reduction) to reduce the number of dimensions to something that is understandable to us humans. Of course, these techniques leave a little bit of information behind, and at times they may be too simplistic. Better than nothing, though, right?

The most famous and used are: **PCA, T-SNE and UMAP**.

```python
from sklearn.decomposition import PCA

# Reduce dimensionality to 3D for visualization
pca = PCA(n_components=3)
word_vecs_3d = pca.fit_transform(word_vecs)
print(f'Explained variance {pca.explained_variance_}')
print(f'Total variance {sum(pca.explained_variance_):.4f}')
```

This code will visualize the word embeddings in a 3D space using PCA for dimensionality reduction. The plot will display each word's position in the 3D space, allowing you to observe the relationships between words in the embedding space.

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create 3D scatter plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(word_vecs_3d[:, 0], word_vecs_3d[:, 1], word_vecs_3d[:, 2], marker='o')

# Add labels to points
for i, word in enumerate(word_vectors.index_to_key):
    ax.text(word_vecs_3d[i, 0], word_vecs_3d[i, 1], word_vecs_3d[i, 2], word)

# Set labels and title
ax.set_xlabel('PCA Component 1')
ax.set_ylabel('PCA Component 2')
ax.set_zlabel('PCA Component 3')
ax.set_title('Word Embeddings Visualization (Word2Vec)')

# Show plot
plt.show()
```

```python
import plotly.graph_objs as go

def plot_3d(dimens, labels, title: str):
    trace = go.Scatter3d(
        x=dimens[:, 0],
        y=dimens[:, 1],
        z=dimens[:, 2],
        mode='markers',
        text=labels,
        marker=dict(
            size=10,
            color='blue',    # set color to an array/list of desired values
            opacity=0.8
        )
    )
    # Create layout
    layout = go.Layout(
        title=title,
        scene=dict(
            xaxis=dict(title='Component 1'),
            yaxis=dict(title='Component 2'),
            zaxis=dict(title='Component 3')
        )
    )
    # Create figure
    fig = go.Figure(data=[trace], layout=layout)
    # Show plot
    fig.show()
```

### Coming back to the embeddings

You can use cosine similarity to compute the similarity between two word embeddings

```python
from scipy.spatial.distance import cosine

# Example words
word1 = 'king'
word2 = 'queen'
word3 = 'car'

embedding1 = model[word1]
embedding2 = model[word2]
embedding3 = model[word3]

# Compute cosine similarity
similarity12 = cosine(embedding1, embedding2)
print(f"Distance between '{word1}' and '{word2}': {similarity12 :.2f}")

similarity23 = cosine(embedding2, embedding3)
print(f"Distance between '{word2}' and '{word3}': {similarity23 :.2f}")
```

# 🤖 Transformers word embedding



Word embeddings from BERT (Bidirectional Encoder Representations from Transformers) offer several advantages over traditional word embeddings from methods like Word2Vec:

-  BERT employs a **bidirectional approach to understand the meaning of words by considering both left and right context**. This enables it to capture more comprehensive semantics compared to the context-independent representations of Word2Vec.

- BERT is **pre-trained on a large corpus of text**, enabling it to learn rich representations of words and their contextual relationships. In contrast, Word2Vec typically requires a large amount of training data to achieve comparable performance.

- BERT **embeddings can be fine-tuned on specific downstream tasks**, allowing them to adapt to the nuances and requirements of different applications. This fine-tuning capability makes BERT embeddings more versatile and applicable across various natural language processing tasks.

- BERT has a **built-in mechanism to handle out-of-vocabulary words** by breaking them down into subwords using WordPiece tokenization.



```python
from transformers import BertTokenizer, BertModel
import torch

# Load pretrained BERT model and tokenizer
model_name = 'bert-base-uncased'
tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertModel.from_pretrained(model_name)
```

#### Let's test the real power of BERT and its context capabilities:

We mentioned that **BERT also considers context when producing embeddings**. How can we test this?

Suppose we have the word "`mouse`" which can be either the animal or the object with which the pointer to the computer is used.

Sentence 1: `The cat chases the mouse`

Sentence 2: `Apple's mouse has a terrible design`

<p align="center">
    <img src="https://i.imgur.com/r9Z22I6.png" alt="drawing" width="300"/>
</p>


**Word2Vec would only be able to provide one embedding of this word, BERT instead returns us two different embeddings**, one for when we are referring to the animal, the second in case we are talking about the device.

```python
# both sentences containing the same word but in two different contexts:
sentence1 = "The cat chases the mouse"
sentence2 = "Apple's mouse has a terrible design"
```

```python
# we want to analyze the word "mouse"
word_of_interest = 'mouse'
token_of_interest = tokenizer.convert_tokens_to_ids(
    tokenizer.tokenize(word_of_interest)
)[0]
print(f'The word "{word_of_interest}" corresponds to the token: {token_of_interest}\n')
```

```python
# Tokenize the input and save the position of the token of interest
ids_sent_1 = tokenizer(sentence1, return_tensors='pt').input_ids
pos_tok_of_interest1 = (ids_sent_1.squeeze() == token_of_interest).nonzero(as_tuple=True)[0].item()   # get the position of "mouse" inside the tokenized sentence
print(f'First sentence tokenized as: {ids_sent_1}')
print(f'We want to get the embedding of the token {token_of_interest} which is in pos. {pos_tok_of_interest1}\n')
```

```python
ids_sent_2 = tokenizer(sentence2, return_tensors='pt').input_ids
pos_tok_of_interest2 = (ids_sent_2.squeeze() == token_of_interest).nonzero(as_tuple=True)[0].item()   # get the position of "mouse" inside the tokenized sentence
print(f'Second sentence tokenized as: {ids_sent_2}')
print(f'We want to get the embedding of the token {token_of_interest} which is in pos. {pos_tok_of_interest2}\n')
```

```python
# Get the model's embeddings for both sentences
with torch.no_grad():
    print('Computing embeddings ...\n')
    output_sent_1 = model(ids_sent_1)
    output_sent_2 = model(ids_sent_2)
print('Done!')

# Extract word embeddings from the output
embeddings1 = output_sent_1.last_hidden_state.squeeze()
embeddings2 = output_sent_2.last_hidden_state.squeeze()

# Get the embedding of the token of interest (the word "mouse")
tok_emb1 = embeddings1[pos_tok_of_interest1]
tok_emb2 = embeddings2[pos_tok_of_interest2]
```

```python
# Are the two embedding of the same token equal?
if torch.equal(tok_emb1, tok_emb2):
    print(f'The token of the word "{word_of_interest}" has the same embedding for both sentences')
else:
    print(f'The token of the word "{word_of_interest}" is embedded in two different ways!')
```

**It worked!**

Although we asked BERT **to embed the same token (i.e., the word mouse)**, the model produced **two completely different embeddings** because the sentence in which they were used implied two different meanings!

#### But wait! We can do even more with BERT!

Remember the special token `[CLS]`? It means `classification` and is used to provide **an embedding of the entire sentence**. This means that we can use a single embedding (sized 768 dimensions) to represent the entire sentence.

```python
# Extract sentence embedding ([CLS] token embedding in 0th position)
sentences = [
    'The movie was a lot of fun',
    'I really enjoyed the movie',
    'I like to go to the theatre',

    'Angular is a javascript framework',
    'Python is a simple programming language',
    'C is a fast programming language',
]

input_ids = tokenizer(sentences, return_tensors='pt', padding=True)

with torch.no_grad():
    output_sent = model(**input_ids)

# get the CLS token (0th dimension)
embeddings_CLS = output_sent.last_hidden_state[:, 0, :]
```

```python
pca = PCA(n_components=3)
embeddings_3d = pca.fit_transform(embeddings_CLS)

plot_3d(embeddings_3d, labels=sentences, title="Sentences embeddings with BERT")
```

```python
word_vecs_3d_TSNE = TSNE(
    n_components=3,
    learning_rate='auto',
    init='random',
    perplexity=5,       # number of nearest neighbors that is used. Must be < samples [5, 50], hint: 30
).fit_transform(embeddings_CLS)
plot_3d(word_vecs_3d_TSNE, labels=sentences, title="Sentences embeddings with BERT w. TSNE")
```

```python
from tqdm import tqdm
def get_embedding(model, tokenized_input, ):
    embeddings = []
    for inp in tqdm(tokenized_input, total=len(tokenized_input)):
        with torch.no_grad():
            output_sent = model(**inp)
        # get the CLS token (0th dimension)
        embeddings.append(output_sent.last_hidden_state[:, 0, :].squeeze())
    return embeddings

tokenized_dataset = []
for sentence in dataset['text']:
    tokenized_dataset.append(tokenizer(sentence, max_length = 512, truncation=True, return_tensors='pt'))

embeddings = get_embedding(model, tokenized_dataset)
```
